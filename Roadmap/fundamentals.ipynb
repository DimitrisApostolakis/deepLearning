```python
float_32_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=torch.float32)
float_32_tensor.device
```

```python
tensor = torch.tensor([1, 2, 3])
print(tensor + 10)
print(tensor.add(10))
```

```python
print(tensor * 10)
print(tensor.mul(10))
```

```python
tensor = tensor - 10
tensor
```

```python
tensor1 = torch.tensor([1, 2, 3])
tensor2 = torch.tensor([4, 5, 6])

matrix_mult = torch.matmul(tensor1, tensor2)
print(matrix_mult)
print(tensor1 @ tensor2)

# tensor1 * tensor2 is element-wise multiplication
element_wise_mult = tensor1 * tensor2
print(element_wise_mult)
```

```python
x = torch.arange(0, 100, 10, dtype=float)
print(x)
print(f"Maximum: {x.max()}")
print(f"Minimum: {x.min()}")
print(f"Mean: {x.mean()}") # Wothout dtype float, it doesn't work
print(f"Sum: {x.sum()}")
```

```python
print(f"Position of max: {x.argmax()}")
print(f"Position of min: {x.argmin()}")
```

```python
print(x.dtype)
tensor_int32 = x.type(torch.int32)
print(tensor_int32.dtype)
```
