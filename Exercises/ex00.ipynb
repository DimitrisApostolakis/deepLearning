```python
import torch

tensor1 = torch.rand(size=(7,7))
tensor2 = torch.rand(size=(1,7))
tensor1, tensor2
```

```python
matrix_mul = tensor1 @ tensor2.T
matrix_mul
```

```python
torch.manual_seed(0)
tensor1 = torch.rand(size=(7,7))
tensor2 = torch.rand(size=(1, 7))

matrix_mul = tensor1 @ tensor2.T
matrix_mul
```

```python
torch.manual_seed(1234)
device = "cuda"

tensor1 = torch.rand(size=(2,3)).to(device)
tensor2 = torch.rand(size=(2,3)).to(device)

tensor1, tensor2
```

```python
matrix_mul = tensor1 @ tensor2.T
matrix_mul
```

```python
print(f"Max of tensor1: {tensor1.max():.2f}")
print(f"Max of tensor2: {tensor2.max():.2f}")

print(f"Min of tensor1: {tensor1.min():.2f}")
print(f"Min of tensor2: {tensor2.min():.2f}")
```

```python
print(f"Index of tensor1's max: {tensor1.argmax()}")
print(f"Index of tensor2's max: {tensor2.argmax()}")

print(f"Index of tensor1's min: {tensor1.argmin()}")
print(f"Index of tensor2's min: {tensor2.argmin()}")
```
